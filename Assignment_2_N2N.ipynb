$$C(w,b) = \frac{1}{2n} \sum_x \|y(x) - a\|^2$$

### Q2a - The Factor of 1/2

Why is there a factor of 1/2 in front? Derive why it is mathematically convenient when computing the derivative of C with respect to any weight w.

> **Your answer:**

The factor 1/2 is included in the squared error cost function to simplify differentiation. When computing the derivative of the cost function with respect to a weight, the power rule produces a factor of 2. The factor 1/2 cancels this 2, resulting in a cleaner gradient expression. This makes the mathematics simpler while leaving the location of the minimum unchanged.

Derivation:

Let the cost function be

C = (1/2)(target - output)^2

Differentiating with respect to a weight w:

dC/dw = (1/2) × 2(target - output) × d(target - output)/dw

The 2 cancels with 1/2:

dC/dw = (target - output) × d(target - output)/dw

Thus, the factor 1/2 is included purely for mathematical convenience.

### Q2c - Cost vs Accuracy

Explain why we use a cost function and do **not** directly maximise classification accuracy.

> **Your answer:**
We do not directly maximize classification accuracy because accuracy is a discrete and non-differentiable measure. Small changes in the network weights often do not change the predicted class, so the accuracy remains unchanged even when the model is improving. As a result, accuracy provides little information about how the weights should be adjusted.

A cost function, on the other hand, is continuous and differentiable. Even small improvements in the network's predictions produce changes in the cost value. This allows gradient descent and backpropagation to compute gradients and systematically update the weights to improve the model.

For example, if the true label is 1, predictions of 0.51 and 0.99 both give 100% accuracy. However, the prediction 0.99 is much better, and the cost function captures this difference while accuracy does not

### Q2d - Absolute Error vs Squared Error

An alternative cost is the absolute error: C = (1/n) * sum |y(x) - a|.

Explain **one mathematical problem** with using absolute error instead of squared error, specifically when doing gradient descent.

> **Your answer:**
A major problem with using absolute error in gradient descent is that the absolute value function is not differentiable at zero. Gradient descent relies on computing derivatives of the cost function with respect to the weights in order to determine the direction of weight updates.

For the absolute error cost, the derivative is different on either side of zero and is undefined when the prediction error is exactly zero. Because of this, the gradient may not exist at some points, making optimization more difficult.

In contrast, the squared error cost is smooth and differentiable everywhere, providing well-defined gradients that can be used efficiently by gradient descent.

The absolute error function has a sharp corner at zero error, so its derivative is undefined there. Since gradient descent requires derivatives, absolute error is mathematically less convenient than squared error.

## Q7 - Geometric Interpretation of 1D Gradient Descent

Consider C(v) as a function of a single scalar v.

1. Give a precise **geometric interpretation** of the update rule v = v - alpha * dC/dv.
2. How do the slope and the learning rate alpha physically dictate movement along the curve?

> **Your answer:**
Geometrically, gradient descent can be viewed as a point moving along the graph of the cost function C(v) in search of the minimum. At any point v, the derivative dC/dv gives the slope of the tangent to the curve. The update rule

v = v - α(dC/dv)

moves the parameter in the direction opposite to the slope because that is the direction of steepest decrease of the cost.

If the slope is positive, the curve rises as v increases, so gradient descent moves left (decreases v). If the slope is negative, the curve falls as v increases, so gradient descent moves right (increases v). When the slope is zero, the algorithm has reached a stationary point and stops moving.

The learning rate α determines the size of each step. A larger α causes larger jumps along the curve and may overshoot the minimum. A smaller α causes smaller, safer steps but makes convergence slower. The magnitude of the slope determines how strongly the parameter is pushed, while α scales that movement.
